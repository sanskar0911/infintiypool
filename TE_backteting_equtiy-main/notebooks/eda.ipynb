# Exploratory Data Analysis (EDA) - Equity Backtesting System

This notebook performs exploratory analysis on processed historical stock price data for major NSE equities:
- RELIANCE
- TCS
- INFY
- HDFCBANK
- ICICIBANK

```python
import os
import glob
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Set display options
pd.set_option('display.max_columns', None)
plt.rcParams['figure.figsize'] = (12, 6)
plt.style.use('seaborn-v0_8-whitegrid') if 'seaborn-v0_8-whitegrid' in plt.style.available else None
```

## 1. Load Processed Stock Data

```python
data_dir = os.path.join('..', 'data', 'processed')
files = glob.glob(os.path.join(data_dir, '*.csv'))

stocks = {}
for f in files:
    ticker = os.path.splitext(os.path.basename(f))[0]
    df = pd.read_csv(f, index_col='Date')
    df.index = pd.to_datetime(df.index)
    stocks[ticker] = df
    print(f"Loaded {ticker}: {df.shape[0]} rows ({df.index.min().date()} to {df.index.max().date()})")
```

## 2. Adjusted Close Price Trends

```python
plt.figure(figsize=(14, 7))
for ticker, df in stocks.items():
    plt.plot(df.index, df['Adj Close'], label=ticker)
plt.title('Normalized Stock Price Performance', fontsize=14, fontweight='bold')
plt.xlabel('Date')
plt.ylabel('Adjusted Close Price (INR)')
plt.legend()
plt.tight_layout()
plt.show()
```

## 3. Daily Returns & Volatility Analysis

```python
returns_df = pd.DataFrame({ticker: df['Returns'] for ticker, df in stocks.items()})
print("Daily Return Summary Statistics:")
display(returns_df.describe())

returns_df.hist(bins=50, figsize=(14, 10), edgecolor='black', alpha=0.7)
plt.suptitle('Daily Return Histograms', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 4. Cross-Asset Correlation Matrix

```python
corr = returns_df.corr()
print("Correlation Matrix:")
display(corr)

plt.figure(figsize=(8, 6))
plt.imshow(corr, cmap='coolwarm', interpolation='none')
plt.colorbar()
plt.xticks(range(len(corr)), corr.columns, rotation=45)
plt.yticks(range(len(corr)), corr.columns)
plt.title('Stock Return Correlation Matrix', fontweight='bold')
plt.tight_layout()
plt.show()
```
