This notebook demonstrates the core Python equity backtesting engine: loading raw historical CSV data, generating technical signals (SMA Crossover), running portfolio simulation with friction, computing quantitative performance KPIs, and comparing returns against the Nifty 50 benchmark.

```python
import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Ensure src directory imports work
sys.path.append(os.path.abspath("../src"))

from data_loader import load_stock_data
from strategy import generate_signals
from backtester import run_backtest
from metrics import calculate_metrics
from benchmark import compare_strategy_vs_benchmark, load_benchmark_data
```

## Step 1: Load Historical Stock Data

```python
ticker = "INFY"
data_path = f"../data/processed/{ticker}.csv"
df = pd.read_csv(data_path, index_col="Date")
df.index = pd.to_datetime(df.index)
print(f"Loaded {len(df)} rows of price data for {ticker}.")
df.tail(5)
```

## Step 2: Generate SMA Crossover Signals

```python
df_signals = generate_signals(df, strategy_name="SMA", short_window=20, long_window=50)
print("Signals generated successfully.")
df_signals[["Close", "SMA_20", "SMA_50", "Signal", "Position"]].tail(10)
```

## Step 3: Run Portfolio Backtest Simulation

```python
portfolio_df, trade_log_df = run_backtest(
    df_signals,
    initial_capital=100000.0,
    commission_pct=0.001,
    slippage_pct=0.0005,
    position_size=0.2,
    stop_loss=0.05,
    take_profit=0.10
)
print(f"Backtest complete. Total trades executed: {len(trade_log_df)}")
portfolio_df.tail(5)
```

## Step 4: Compute Quantitative Metrics

```python
metrics = calculate_metrics(portfolio_df, trade_log_df, initial_capital=100000.0)
print("--- Quantitative Metrics Output ---")
for k, v in metrics.items():
    if isinstance(v, float):
        print(f"{k:25s}: {v:.4f}")
    else:
        print(f"{k:25s}: {v}")
```

## Step 5: Compare Against Nifty 50 Benchmark

```python
benchmark_df = load_benchmark_data("NIFTY50")
bench_comp = compare_strategy_vs_benchmark(portfolio_df, benchmark_df, benchmark_symbol="NIFTY50")
print("--- Benchmark Comparison ---")
for k, v in bench_comp.items():
    print(f"{k:20s}: {v}")
```
